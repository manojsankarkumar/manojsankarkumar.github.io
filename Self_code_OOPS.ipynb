```python
# ============================================================
# CONCEPT 0 — WHAT IS OOP?
# ============================================================
# OOPS = Object-Oriented Programming Structure
# Instead of writing a big pile of functions and variables,
# we group related data + behaviour into "objects".
#
# Real-world analogy:
#   A bank has many accounts. Each account has:
#     - Data    → owner name, PIN, balance
#     - Actions → deposit, withdraw, check balance
#
# We build one BLUEPRINT (class) and stamp out as many
# individual OBJECTS (accounts) as we need.
# ============================================================

# ============================================================
# CONCEPT 1 — CLASS & OBJECT
# ============================================================
# CLASS  = the blueprint / cookie cutter
# OBJECT = a real thing built from that blueprint / a cookie
#
# __init__  → the "setup" method. Runs automatically when you
#             create a new object.
# self      → means "this particular object". When alice calls
#             a method, self IS alice.
# ============================================================

print("="*50)
print("CONCEPT 1 - CLASS AND OBJECT")
print("="*50)

class BankAccount():
  def __init__(self,owner_name,pin):
    self.owner_name = owner_name  #public: anyone can read
    self.pin = pin                # to be made private, as this is confidential
    self.balance = 0              # balance initated to 0 here
    print(f"Account created for {self.owner_name}")

  def show_info(self):
    #Printing the summary of the account created"
    print(f" Owner: {self.owner_name} | Balance: {self.balance}")

# Creating objects from the class

alice = BankAccount("Alice","1234") #Object 1 - Account created for Alice
bob = BankAccount("Bob","5678")     #Object 2 - Account created for Bob
carol = BankAccount("Carol", "9999") #Object 3 - Account created for Carol

# Each object is completely separate even though they share the
# same blueprint — changing alice does NOT affect bob or carol.
alice.balance = 500
bob.balance = 1000


alice.show_info() # Owner: Alice | Balance: 500
bob.show_info() # Owner: Bob | Balance: 1000
carol.show_info() # Owner: Carol | Balance: 0

print()
print("  Key facts:")
print(f"  type(alice) → {type(alice)}")
print(f"  alice is bob → {alice is bob}")   # False: different objects
print()
```

###### 2. ENCAPSULATION

```python
# ============================================================
# CONCEPT 2 — ENCAPSULATION
# ============================================================
# Encapsulation = hiding private data + providing safe doorways
#
# Why? So nobody can accidentally (or maliciously) break your
# object by messing with its internals directly.
#
# Python rule: put __ (two underscores) BEFORE a name to make
# it private. Python renames it to _ClassName__name internally
# (this is called "name mangling").
#
# Private:  self.__pin, self.__balance   ← locked away
# Public:   self.owner_name              ← anyone can read
#
# Doorways (methods) are the only safe way to interact with
# private data. They can validate input, check PINs, log
# activity, etc. before touching the data.
# ============================================================

print("=" * 50)
print("  CONCEPT 2 — ENCAPSULATION")
print("=" * 50)

class BankAccount_V2:
  def __init__(self,owner_name,pin):
    self.owner_name = owner_name
    self.__pin = pin
    self.__balance = 0
    print(f"Account created for {self.owner_name}")

# Doorway 1: Need to check the pin as pre-requisite before any operations to be done in bank

  def check_pin(self,entered_pin):
    if entered_pin == self.__pin:
      return True
    else:
      print(f"\n Incorrect Pin on {self.owner_name} login")
      return False

# Doorway 2: Display the account balance here when login successful

  def get_balance(self):
    return self.__balance #self.__balance is created in __init__ method above

# Doorway 3: Deposit money

  def deposit_money(self,amount,entered_pin):
    """Add money - but only with correct PIN"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Deposit amount must be greater than zero!")
      return
    self.__balance += amount
    print(f"\n Deposited ${amount}. New balance: ${self.__balance} on {self.owner_name} account")

# Doorway 4: Withdraw money

  def withdraw_money(self,amount,entered_pin):
    """Remove Money - only if the pin entered is correct"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Withdraw amount must be greater than zero!")
      return
    if amount > self.__balance:
      print(f"Insufficient funds. Balance: ${self.__balance}")
      return
    self.__balance -= amount
    print(f"\n Withdrawn ${amount}. New balance: ${self.__balance}")

# Doorway 5: Change PIN

  def change_pin(self,old_pin,new_pin):
    """Change the PIN - only if the old PIN is correct"""
    if not self.check_pin(old_pin):
      return
    self.__pin = new_pin
    print("PIN changed successfully!")

  def __str__(self):
    return f"Bank Account(owner:'{self.owner_name}', balance=${self.__balance:.2f})"

alice = BankAccount_V2("Alice", "1234")

# Trying to access private data directly
print("\n Trying alice.__balance")
try:
  print(alice.__balance)
except AttributeError as e:
  print(f" Attribute Error: {e}")

# Using the safe doorway -> WORKS
print(f"\n alice.get_balance() -> ${alice.get_balance()}")

# Depositing $500 with correct PIN
print("\n Depositing $500 with correct PIN")
alice.deposit_money(500, "1234")

print("\n Withdrawing $200 with incorrect PIN")
alice.withdraw_money(200, "9999")

print("\n  Withdrawing $200 with correct PIN:")
alice.withdraw_money(200, "1234")

print("\n  Changing PIN:")
alice.change_pin("1234", "4321")

#def __str__(self) -> used below
print(f"\n print(alice) -> {alice}")
```

###### 3. ABSTRACTION

Abstraction: Introduces AbstractAccount(ABC) — a pure contract with @abstractmethod. Shows the TypeError you get if you try to instantiate it directly. Then BankAccount_v3 properly implements all three required methods. Abstraction says what must exist; the concrete class decides how.

```python
# ============================================================
# CONCEPT 3 — ABSTRACTION
# ============================================================
# Abstraction = showing WHAT something does, hiding HOW it does it
#
# Think of a TV remote. You press "Volume Up" and the volume goes
# up. You don't know (or care) about the infrared signals, the
# circuit board, or the firmware. The remote ABSTRACTS all that.
#
# In Python we use the abc module (Abstract Base Class):
#   - ABC        → marks a class as abstract
#   - @abstractmethod → marks a method that EVERY child MUST implement
#
# An abstract class CANNOT be used directly — it's just a contract
# that says "any class that inherits me must provide these methods".
#
# AbstractAccount below forces ALL account types to have:
#   deposit_money(), withdraw(), get_balance()
# If a child forgets even one, Python raises a TypeError instantly.
# ============================================================
print("=" * 55)
print("  CONCEPT 3 — ABSTRACTION")
print("=" * 55)

from abc import ABC, abstractmethod

class AbstractAccount(ABC):

  @abstractmethod
  def deposit_money(self,amount,entered_pin):
    """Deposit funds into the account"""
    pass # no code here - the child fills in

  @abstractmethod
  def withdraw_money(self,amount,entered_pin):
    """Withdraw funds from the account"""
    pass

  @abstractmethod
  def get_balance(self):
    """Get the balance of the account"""
    pass

# Note: a class can mix abstract AND concrete methods.
# This is a concrete method — shared by every child for free.

  def account_type_info(self):
    print(f"This is a {type(self).__name__} account")

#Trying to instantiate the abstract class -> TypeError
print("Trying to create AbstractAccount directly:")
try:
  acc = AbstractAccount()
except TypeError as e:
  print(f" TypeError - {e}")


# A class that PROPERLY implements all abstract methods

class BankAccount_V3(AbstractAccount):
  """Concrete Account - fulfills the AbstractAccount contract"""

  def __init__(self,owner_name,pin):
    self.owner_name = owner_name
    self.__pin = pin
    self.__balance = 0
    print(f"Account created for {self.owner_name}")

# Doorway 1: Need to check the pin as pre-requisite before any operations to be done in bank

  def check_pin(self,entered_pin):
    if entered_pin == self.__pin:
      return True
    else:
      print(f"\n Incorrect Pin on {self.owner_name} login")
      return False

  # Doorway 2: Display the account balance here when login successful

  def get_balance(self):                                                     # <--implements abstract method
    return self.__balance #self.__balance is created in __init__ method above

# Doorway 3: Deposit money

  def deposit_money(self,amount,entered_pin):                               # <--implements abstract method
    """Add money - but only with correct PIN"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Deposit amount must be greater than zero!")
      return
    self.__balance += amount
    print(f"\n Deposited ${amount}. New balance: ${self.__balance} on {self.owner_name} account")

# Doorway 4: Withdraw money

  def withdraw_money(self,amount,entered_pin):                             # <--implements abstract method
    """Remove Money - only if the pin entered is correct"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Withdraw amount must be greater than zero!")
      return
    if amount > self.__balance:
      print(f"Insufficient funds. Balance: ${self.__balance}")
      return
    self.__balance -= amount
    print(f"\n Withdrawn ${amount}. New balance: ${self.__balance}")

  # Doorway 5: Change PIN

  def change_pin(self,old_pin,new_pin):
    """Change the PIN - only if the old PIN is correct"""
    if not self.check_pin(old_pin):
      return
    self.__pin = new_pin
    print("PIN changed successfully!")

  def __str__(self):
    return f"Bank Account(owner:'{self.owner_name}', balance=${self.__balance:.2f})"

print("\n Creating a BankAccount_V3:")
alice = BankAccount_V3("Alice", "1234")

alice.account_type_info()                     # inherited concrete method from abstract class
alice.deposit_money(300,"1234")

print(f" Balance: ${alice.get_balance()}")
print()
```

###### 4. INHERITANCE

 Inheritance: SavingsAccount(BankAccount_v3) inherits everything for free and adds only add_interest(). Shows super().__init__() and the full method resolution chain.

```python
# ============================================================
# CONCEPT 4 — INHERITANCE
# ============================================================
# Inheritance = a child class automatically gets all of the
# parent's methods and attributes, then adds its own.
#
# Why? Avoids copy-pasting the same code everywhere.
#
# Syntax:  class Child(Parent):
# super()  → lets the child call the parent's methods
# super().__init__() → run parent's setup before child's own
#
# Here BankAccount_v3 inherits AbstractAccount (the contract).
# SavingsAccount and PremiumAccount both inherit BankAccount_v3
# (the working implementation).
#
# Hierarchy:
#   AbstractAccount
#       └── BankAccount_v3
#               ├── SavingsAccount   (adds: add_interest)
#               └── PremiumAccount   (adds: cashback on withdrawal)
# ============================================================

print("=" * 55)
print("  CONCEPT 4 — INHERITANCE")
print("=" * 55)

class SavingsAccount(BankAccount_V3):        #Adding a class as an attribute will inherit. This is Inheritance
  """Inherits everything from BankAccount_v3. Adds interest earning capability on top."""
  interest_rate = 0.05

  def __init__(self, owner_name, pin):
    super().__init__(owner_name, pin)   # ← run parent's __init__ first!
    print(f" Savings Account - earns{self.interest_rate*100:.0f}% interest.")

# NEW method only Savings Account has
  def add_interest(self):
    """Credit Interest to the balance"""
    interest = self.get_balance() * self.interest_rate
    # __balance is private to BankAccount_v3, so we use the
    # mangled name to reach it from a grandchild class.
    self._BankAccount_V3__balance += interest
    print(f" Interest ${interest:.2f} added. "
          f" New balance: ${self.get_balance():.2f}")

  def __str__(self):
    return (f"SavingsAccount(owner='{self.owner_name}', "
                f"balance=${self.get_balance():.2f})")


print("\n  Creating Bob as a SavingsAccount:")
bob = SavingsAccount("Bob", "5678")

# Bob has ALL of BankAccount_v3's methods (inherited)...
bob.deposit_money(1000, "5678")

# ...PLUS his own add_interest() method
bob.add_interest()

bob.withdraw_money(200, "5678")
print(f"\n  {bob}")
print()
```
