```python
# ============================================================
# CONCEPT 0 — WHAT IS OOP?
# ============================================================
# OOPS = Object-Oriented Programming Structure
# Instead of writing a big pile of functions and variables,
# we group related data + behaviour into "objects".
#
# Real-world analogy:
#   A bank has many accounts. Each account has:
#     - Data    → owner name, PIN, balance
#     - Actions → deposit, withdraw, check balance
#
# We build one BLUEPRINT (class) and stamp out as many
# individual OBJECTS (accounts) as we need.
# ============================================================

# ============================================================
# CONCEPT 1 — CLASS & OBJECT
# ============================================================
# CLASS  = the blueprint / cookie cutter
# OBJECT = a real thing built from that blueprint / a cookie
#
# __init__  → the "setup" method. Runs automatically when you
#             create a new object.
# self      → means "this particular object". When alice calls
#             a method, self IS alice.
# ============================================================

print("="*50)
print("CONCEPT 1 - CLASS AND OBJECT")
print("="*50)

class BankAccount():
  def __init__(self,owner_name,pin):
    self.owner_name = owner_name  #public: anyone can read
    self.pin = pin                # to be made private, as this is confidential
    self.balance = 0              # balance initated to 0 here
    print(f"Account created for {self.owner_name}")

  def show_info(self):
    #Printing the summary of the account created"
    print(f" Owner: {self.owner_name} | Balance: {self.balance}")

# Creating objects from the class

alice = BankAccount("Alice","1234") #Object 1 - Account created for Alice
bob = BankAccount("Bob","5678")     #Object 2 - Account created for Bob
carol = BankAccount("Carol", "9999") #Object 3 - Account created for Carol

# Each object is completely separate even though they share the
# same blueprint — changing alice does NOT affect bob or carol.
alice.balance = 500
bob.balance = 1000


alice.show_info() # Owner: Alice | Balance: 500
bob.show_info() # Owner: Bob | Balance: 1000
carol.show_info() # Owner: Carol | Balance: 0

print()
print("  Key facts:")
print(f"  type(alice) → {type(alice)}")
print(f"  alice is bob → {alice is bob}")   # False: different objects
print()
```

```python
# ============================================================
# CONCEPT 2 — ENCAPSULATION
# ============================================================
# Encapsulation = hiding private data + providing safe doorways
#
# Why? So nobody can accidentally (or maliciously) break your
# object by messing with its internals directly.
#
# Python rule: put __ (two underscores) BEFORE a name to make
# it private. Python renames it to _ClassName__name internally
# (this is called "name mangling").
#
# Private:  self.__pin, self.__balance   ← locked away
# Public:   self.owner_name              ← anyone can read
#
# Doorways (methods) are the only safe way to interact with
# private data. They can validate input, check PINs, log
# activity, etc. before touching the data.
# ============================================================

print("=" * 50)
print("  CONCEPT 2 — ENCAPSULATION")
print("=" * 50)

class BankAccount_V2:
  def __init__(self,owner_name,pin):
    self.owner_name = owner_name
    self.__pin = pin
    self.__balance = 0
    print(f"Account created for {self.owner_name}")

# Doorway 1: Need to check the pin as pre-requisite before any operations to be done in bank
  def check_pin(self,entered_pin):
    if entered_pin == self.__pin:
      return True
    else:
      print(f"\n Incorrect Pin on {self.owner_name} login")
      return False

# Doorway 2: Display the account balance here when login successful
  def get_balance(self):
    return self.__balance #self.__balance is created in __init__ method above

# Doorway 3: Deposit money
  def deposit_money(self,amount,entered_pin):
    """Add money - but only with correct PIN"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Deposit amount must be greater than zero!")
      return
    self.__balance += amount
    print(f"\n Deposited ${amount}. New balance: ${self.__balance} on {self.owner_name} account")

# Doorway 4: Withdraw money
  def withdraw_money(self,amount,entered_pin):
    """Remove Money - only if the pin entered is correct"""
    if not self.check_pin(entered_pin):
      return
    if amount <= 0:
      print("Withdraw amount must be greater than zero!")
      return
    if amount > self.__balance:
      print(f"Insufficient funds. Balance: ${self.__balance}")
      return
    self.__balance -= amount
    print(f"\n Withdrawn ${amount}. New balance: ${self.__balance}")

# Doorway 5: Change PIN

  def change_pin(self,old_pin,new_pin):
    """Change the PIN - only if the old PIN is correct"""
    if not self.check_pin(old_pin):
      return
    self.__pin = new_pin
    print("PIN changed successfully!")

  def __str__(self):
    return f"Bank Account(owner:'{self.owner_name}', balance=${self.__balance:.2f})"

alice = BankAccount_V2("Alice", "1234")

# Trying to access private data directly
print("\n Trying alice.__balance")
try:
  print(alice.__balance)
except AttributeError as e:
  print(f" Attribute Error: {e}")

# Using the safe doorway -> WORKS
print(f"\n alice.get_balance() -> ${alice.get_balance()}")

# Depositing $500 with correct PIN
print("\n Depositing $500 with correct PIN")
alice.deposit_money(500, "1234")

print("\n Withdrawing $200 with incorrect PIN")
alice.withdraw_money(200, "9999")

print("\n  Withdrawing $200 with correct PIN:")
alice.withdraw_money(200, "1234")

print("\n  Changing PIN:")
alice.change_pin("1234", "4321")
```
