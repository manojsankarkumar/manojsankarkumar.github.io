```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.get_dataset_names()

penguin = sns.load_dataset('penguins')
penguin
```

```python
#Step 1: Understanding the dataset

penguin.dtypes

penguin.isna().sum()
```

```python
#Step 2: Cleaning dataset
# Since the penguin dataset has NA in few columns, this should be dropped
penguin_clean_data = penguin.dropna()
penguin_clean_data
```

#### 1. Univariate Analysis

##### 1.1 Histogram

```python
plt.hist(data=penguin_clean_data,x="flipper_length_mm",color="lightgreen")
plt.xlabel("Flipper Length(mm)")
plt.show()
```

```python
sns.histplot(data=penguin_clean_data,y="bill_depth_mm",color="lightblue")
plt.ylabel("Bill Depth(mm)")
plt.show()
```

##### 1.2 Countplot

```python
sns.countplot(data=penguin_clean_data, x="species")
plt.xlabel("SPECIES")
plt.ylabel("COUNT")
plt.show()
```

```python
sns.countplot(data=penguin_clean_data, x="island",hue="species")
plt.xlabel("ISLAND")
plt.ylabel("COUNT")
plt.show()
```

##### 1.3 Piechart



```python
plt.pie(x=penguin_clean_data["island"].value_counts(),labels=["Adelie","Chinstrap","Gentoo"],explode=[0.01,0.01,0.01],autopct='%1.0f%%')
plt.show()
```

```python
plt.pie(x=penguin_clean_data["species"].value_counts(),explode=[0.01,0.01,0.01],autopct='%1.0f%%')
plt.show()
```

#### 2. Bivariate Analysis

###### 2.1 Barplot - Seaborn

```python
sns.barplot(data=penguin_clean_data,x="species",y="body_mass_g")
plt.xlabel("Species")
plt.ylabel("Body Mass - G")
plt.show()
```

###### 2.1 Barplot - Matplotlib

```python
plt.bar(data=penguin_clean_data,x="species",height="bill_length_mm")
plt.title("Bill Length by Species")
plt.ylabel("Length of Bill (mm)")
plt.show()
```

```python
plt.bar(data=penguin_clean_data,x="island",height="bill_length_mm")
plt.title("Bill Length by Island")
plt.ylabel("Length of Bill (mm)")
plt.show()
```

###### 2.2 Scatterplot - Seaborn

```python
plt.scatter(data=penguin_clean_data, x="flipper_length_mm", y="bill_length_mm",)
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Bill Length (mm)")
plt.show()
```

```python
sns.scatterplot(data=penguin_clean_data, x="flipper_length_mm", y="bill_length_mm", hue="species")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Bill Length (mm)")
plt.show()
```

#### 3. Multivariate Analysis

###### 3.1 Pairplot - Seaborn

```python
sns.pairplot(data=penguin_clean_data, hue="species")
plt.show()
```

###### 3.2 Heatmap

```python
numerical_penguin = penguin_clean_data.select_dtypes(include=['float64', 'int64'])

corr_matrix = numerical_penguin.corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Penguins")
plt.show()
```
